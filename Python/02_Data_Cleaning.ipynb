```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/raw/games.csv")
```

```python
steam = df.copy()
```

```python
steam.columns = ['Name', 'Release date', 'Estimated owners', 'Peak CCU',
       'Required age', 'Price', 'Discount', 'DLC count', 'About the game',
       'Supported languages', 'Full audio languages', 'Reviews',
       'Header image', 'Website', 'Support url', 'Support email', 'Windows',
       'Mac', 'Linux', 'Metacritic score', 'Metacritic url', 'User score',
       'Positive', 'Negative', 'Score rank', 'Achievements', 'Recommendations',
       'Notes', 'Average playtime forever', 'Average playtime two weeks',
       'Median playtime forever', 'Median playtime two weeks', 'Developers',
       'Publishers', 'Categories', 'Genres', 'Tags', 'Screenshots', 'Movies']
```

```python
steam.drop_duplicates(inplace=True)
```

```python
steam["Release date"] = pd.to_datetime(
    steam["Release date"],
    format="mixed",
    errors="coerce"
)
```

```python
steam.dropna(subset=["Name"], inplace=True)
```

```python
columns_to_drop = [
    "Movies",
    "Score rank",
    "Metacritic url",
    "Reviews",
    "Notes",
    "Website",
    "Support url",
    "Support email",
    "Header image",
    "Screenshots"
]

steam.drop(columns=columns_to_drop, inplace=True)
```

```python
steam["Total Reviews"] = (
    steam["Positive"] +
    steam["Negative"]
)
```

```python
steam["Review Score %"] = np.where(
    steam["Total Reviews"] == 0,
    0,
    (steam["Positive"] / steam["Total Reviews"]) * 100
)
```

```python
steam["Release Year"] = (
    steam["Release date"].dt.year
)
```

```python
steam["Release Month"] = (
    steam["Release date"].dt.month_name()
)
```

```python
def categorize_price(price):

    if price == 0:
        return "Free"

    elif price < 10:
        return "Budget"

    elif price < 30:
        return "Standard"

    elif price < 60:
        return "Premium"

    else:
        return "Expensive"

steam["Price Category"] = (
    steam["Price"].apply(categorize_price)
)
```

```python
steam.to_csv("../Data/Cleaned/steam.csv")
```

```python
genres = steam.copy()
```

```python
genres["Genres"] = genres["Genres"].str.split(",")
```

```python
genres = genres.explode("Genres")
```

```python
genre_counts = (
    genres["Genres"]
    .value_counts()
    .head(15)
)

genre_counts
```

```python
plt.figure(figsize=(12,6))

genre_counts.plot(kind="bar")

plt.title("Top 15 Genres on Steam")
plt.xlabel("Genre")
plt.ylabel("Number of Games")

plt.xticks(rotation=45)

plt.savefig("../Images/Genre Analysis/top_genres.png", dpi=300, bbox_inches="tight")

plt.show()
```
